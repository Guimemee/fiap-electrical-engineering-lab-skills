# Checkpoint 6: Dimensionamento de Transistor BJT NPN como Chave Digital
Cálculo de polarização na saturação e proteção indutiva de comutação.


```python
v_cc = 12.0
v_in = 5.0
r_carga = 120.0
beta_min = 100.0
k_sat = 3.0
v_be_sat = 0.7
v_ce_sat = 0.2

i_c = (v_cc - v_ce_sat) / r_carga
i_b_sat = (i_c / beta_min) * k_sat
r_b_calc = (v_in - v_be_sat) / i_b_sat

print(f'Corrente de Carga Ic:    {i_c*1000:.2f} mA')
print(f'Corrente de Base Ib:     {i_b_sat*1000:.2f} mA')
print(f'Resistor de Base Rb:     {r_b_calc:.1f} Ω (Usar 1.2 kΩ ou 1.5 kΩ)')
print(f'Potência em Rb:          {(i_b_sat**2 * r_b_calc)*1000:.2f} mW')

```

```python
import matplotlib.pyplot as plt
import numpy as np

# Reta de carga do transistor BJT
vce_vals = np.linspace(0, 12, 100)
ic_vals = (12.0 - vce_vals) / 120.0 * 1000.0 # mA

fig, ax = plt.subplots(figsize=(8, 4.5))
ax.plot(vce_vals, ic_vals, 'b-', lw=2.5, label='Reta de Carga CC (Vcc=12V, Rc=120Ω)')
ax.scatter([0.2], [98.33], color='green', s=120, zorder=5, label='Ponto Q de Saturação (Vce=0.2V, Ic=98.3mA)')
ax.scatter([12.0], [0.0], color='red', s=120, zorder=5, label='Ponto Q de Corte (Vce=12V, Ic=0mA)')

ax.set_xlabel('Tensão Coletor-Emissor Vce (V)', fontweight='bold')
ax.set_ylabel('Corrente de Coletor Ic (mA)', fontweight='bold')
ax.set_title('Reta de Carga e Pontos de Operação da Chave com Transistor BJT', fontweight='bold')
ax.legend(loc='upper right')
ax.grid(True, alpha=0.5)
plt.tight_layout()
plt.show()

```
