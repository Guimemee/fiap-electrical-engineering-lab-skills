# Checkpoint 1: Efeito de Carga do Multímetro em Circuitos de Alta Impedância
Cálculo do erro de inserção na medição voltimétrica.


```python
vs = 12.0
r1 = 1e6
r2 = 1e6
rm = 10e6 # Resistencia do voltimetro

v_ideal = vs * (r2 / (r1 + r2))
r_eq = (r2 * rm) / (r2 + rm)
v_med = vs * (r_eq / (r1 + r_eq))
erro_pct = abs(v_med - v_ideal) / v_ideal * 100.0

print(f'Tensão Ideal:          {v_ideal:.2f} V')
print(f'Tensão Medida Real:    {v_med:.2f} V')
print(f'Erro de Inserção:      {erro_pct:.2f}%')

```

```python
import matplotlib.pyplot as plt
import numpy as np

rm_range = np.logspace(5, 8, 100) # 100k a 100M
v_med_vals = [12.0 * (((1e6 * rm)/(1e6 + rm)) / (1e6 + ((1e6 * rm)/(1e6 + rm)))) for rm in rm_range]

fig, ax = plt.subplots(figsize=(8, 4.5))
ax.plot(rm_range / 1e6, v_med_vals, 'b-', lw=2.5, label='Tensão Medida Vout (V)')
ax.axhline(6.0, color='r', linestyle='--', label='Tensão Teórica Ideal (6.0 V)')
ax.scatter([10], [5.71], color='darkgreen', s=90, zorder=5, label='DMM Típico (Rm = 10 MΩ -> 5.71 V)')

ax.set_xscale('log')
ax.set_xlabel('Resistência Interna do Multímetro (MΩ)', fontweight='bold')
ax.set_ylabel('Tensão Medida (V)', fontweight='bold')
ax.set_title('Efeito de Carga do Voltímetro em Função da Resistência Interna', fontweight='bold')
ax.legend(loc='lower right')
ax.grid(True, alpha=0.5)
plt.tight_layout()
plt.show()

```
