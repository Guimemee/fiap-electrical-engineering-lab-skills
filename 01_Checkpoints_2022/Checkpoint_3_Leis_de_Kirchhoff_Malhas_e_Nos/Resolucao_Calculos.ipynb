# Checkpoint 3: Resolução de Malhas de Kirchhoff por Álgebra Matricial
Solução exata do sistema linear das Leis de Kirchhoff com NumPy.


```python
import numpy as np

# Matriz de resistências R e vetor de fontes V:
# [ 10  -6 ] [ I1 ] = [  20 ]
# [ -6   8 ] [ I2 ]   [ -12 ]
R_mat = np.array([
    [ 10.0, -6.0],
    [ -6.0,  8.0]
])
V_vec = np.array([20.0, -12.0])

I_sol = np.linalg.solve(R_mat, V_vec)
i1, i2 = I_sol[0], I_sol[1]
i3 = i1 - i2

print(f'Corrente de Malha I1: {i1:.3f} A')
print(f'Corrente de Malha I2: {i2:.3f} A')
print(f'Corrente Ramo Comum I3: {i3:.3f} A')

```

```python
import matplotlib.pyplot as plt

ramos = ['Ramo 1 (I1)', 'Ramo Central (I3)', 'Ramo 2 (I2)']
correntes = [i1, i3, i2]

fig, ax = plt.subplots(figsize=(7, 4))
bars = ax.bar(ramos, correntes, color=['#1f77b4', '#2ca02c', '#d62728'], width=0.5)
ax.set_ylabel('Intensidade de Corrente Elétrica (A)', fontweight='bold')
ax.set_title('Distribuição de Correntes de Kirchhoff no Circuito Multimalhas', fontweight='bold')
for b in bars:
    h = b.get_height()
    ax.text(b.get_x() + b.get_width()/2.0, h + 0.05, f'{h:.2f} A', ha='center', fontweight='bold')
ax.set_ylim(-0.5, 2.5)
ax.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

```
