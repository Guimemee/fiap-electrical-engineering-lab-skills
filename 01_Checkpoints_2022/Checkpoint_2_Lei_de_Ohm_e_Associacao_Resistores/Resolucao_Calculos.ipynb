# Checkpoint 2: Análise da Lei de Ohm e Potência em Associação Mista
Cálculo de correntes de ramo, quedas de tensão e dissipação térmica Joule.


```python
v_fonte = 24.0
r1 = 100.0
r2 = 300.0
r3 = 600.0

r_p = (r2 * r3) / (r2 + r3)
r_eq = r1 + r_p
i_tot = v_fonte / r_eq

v_r1 = i_tot * r1
p_r1 = i_tot * v_r1

v_p = v_fonte - v_r1
i_r2 = v_p / r2
p_r2 = v_p * i_r2

i_r3 = v_p / r3
p_r3 = v_p * i_r3

print(f'Resistência Total Req: {r_eq:.2f} Ω')
print(f'Corrente Total:        {i_tot*1000:.2f} mA')
print(f'Potência em R1 (100Ω): {p_r1:.3f} W')
print(f'Potência em R2 (300Ω): {p_r2:.3f} W')
print(f'Potência em R3 (600Ω): {p_r3:.3f} W')
print(f'Potência Total:        {(p_r1+p_r2+p_r3):.3f} W')

```

```python
import matplotlib.pyplot as plt

resistores = ['R1 (100 Ω)', 'R2 (300 Ω)', 'R3 (600 Ω)']
potencias = [p_r1, p_r2, p_r3]

fig, ax = plt.subplots(figsize=(7, 4))
bars = ax.bar(resistores, potencias, color=['#d62728', '#ff7f0e', '#1f77b4'], width=0.5)
ax.axhline(0.25, color='gray', linestyle='--', label='Limite Resistor 1/4 W (0.25 W)')
ax.set_ylabel('Potência Dissipada (Watts)', fontweight='bold')
ax.set_title('Dissipação de Potência por Resistor e Verificação de Sobrecarga', fontweight='bold')
for b in bars:
    h = b.get_height()
    ax.text(b.get_x() + b.get_width()/2.0, h + 0.02, f'{h:.3f} W', ha='center', fontweight='bold')
ax.legend()
ax.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

```
