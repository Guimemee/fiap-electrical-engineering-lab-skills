# Checkpoint 5: Dimensionamento de Retificador em Ponte e Filtro Capacitivo
Cálculo de forma de onda retificada, condução dos diodos e capacitor de ripple.


```python
import numpy as np

v_rms = 12.0
f_rede = 60.0
i_dc = 0.500
v_rip_alvo = 0.80

v_p_sec = v_rms * np.sqrt(2)
v_p_dc = v_p_sec - 1.4 # Queda de 2 diodos
f_rip = 2 * f_rede
c_filtro = i_dc / (f_rip * v_rip_alvo)
v_dc_medio = v_p_dc - v_rip_alvo / 2.0

print(f'Tensão de Pico do Secundário: {v_p_sec:.2f} V')
print(f'Tensão de Pico CC na Carga:   {v_p_dc:.2f} V')
print(f'Capacitância Mínima de Filtro: {c_filtro*1e6:.1f} µF')
print(f'Tensão Contínua Média Vdc:     {v_dc_medio:.2f} V')

```

```python
import matplotlib.pyplot as plt
import numpy as np

t = np.linspace(0, 0.0333, 500) # 2 ciclos de 60Hz
v_seno = 15.57 * np.abs(np.sin(2 * np.pi * 60 * t))
# Modelo simplificado com descarga de capacitor
v_filtrada = np.maximum(v_seno, 15.57 - 0.8 * (t % (1/120)) * 120)

fig, ax = plt.subplots(figsize=(8, 4.5))
ax.plot(t * 1000, v_seno, 'gray', linestyle=':', label='Onda Completa Retificada sem Filtro')
ax.plot(t * 1000, v_filtrada, 'b-', lw=2.5, label='Tensão CC Filtrada (Ripple = 0.8 Vpp)')
ax.axhline(15.17, color='g', linestyle='--', label='Tensão Contínua Média Vdc = 15.17 V')

ax.set_xlabel('Tempo (milissegundos)', fontweight='bold')
ax.set_ylabel('Tensão (Volts)', fontweight='bold')
ax.set_title('Forma de Onda de Saída do Retificador com Filtro Capacitivo', fontweight='bold')
ax.legend(loc='lower right')
ax.grid(True, alpha=0.5)
plt.tight_layout()
plt.show()

```
