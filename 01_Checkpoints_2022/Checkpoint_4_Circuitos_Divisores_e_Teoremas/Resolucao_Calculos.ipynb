# Checkpoint 4: Teorema da Máxima Transferência de Potência (Thévenin)
Curva de potência entregue em função da resistência de carga RL.


```python
import numpy as np

v_th = 10.0 # V
r_th = 50.0 # Ohms

rl_range = np.linspace(1, 200, 300)
pot_l = (v_th**2 * rl_range) / ((r_th + rl_range)**2)

idx_max = np.argmax(pot_l)
print(f'Resistência de Thévenin: Rth = {r_th:.1f} Ω')
print(f'RL Ótimo:                 RL* = {rl_range[idx_max]:.1f} Ω')
print(f'Potência Máxima:          Pmax = {pot_l[idx_max]*1000:.1f} mW')

```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 4.5))
ax.plot(rl_range, pot_l * 1000, 'b-', lw=2.5, label='Potência na Carga PL (mW)')
ax.axvline(50, color='r', linestyle='--', label='Condição Ótima RL = Rth = 50 Ω')
ax.scatter([50], [500], color='darkred', s=100, zorder=5, label='Máximo Absoluto (50 Ω, 500 mW)')

ax.set_xlabel('Resistência de Carga RL (Ω)', fontweight='bold')
ax.set_ylabel('Potência Entregue (mW)', fontweight='bold')
ax.set_title('Teorema da Máxima Transferência de Potência (Vth = 10V, Rth = 50Ω)', fontweight='bold')
ax.legend(loc='upper right')
ax.grid(True, alpha=0.5)
plt.tight_layout()
plt.show()

```
