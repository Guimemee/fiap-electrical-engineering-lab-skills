```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Validação do Exercício 8 (Capacitores em Série)
C = np.array([200e-6, 50e-6, 10e-6])
E = 60.0
inv_Ct = np.sum(1.0 / C)
Ct = 1.0 / inv_Ct
Qt = Ct * E
V = Qt / C

print(f'Exercício 8:')
print(f'Capacitância Total Ct: {Ct*1e6:.2f} µF')
print(f'Carga Total Qt:        {Qt*1e6:.2f} µC')
for i, v in enumerate(V):
    print(f'Tensão V{i+1}:             {v:.2f} V')
print(f'Soma das Tensões:      {np.sum(V):.2f} V (Confirmado = E)\n')

# 2. Validação do Exercício 9 (Capacitores em Paralelo)
Cp = np.array([800e-6, 60e-6, 1200e-6])
Ep = 48.0
Ct_paralelo = np.sum(Cp)
Q_indiv = Cp * Ep
Qt_paralelo = np.sum(Q_indiv)

print(f'Exercício 9:')
print(f'Capacitância Total Ct: {Ct_paralelo*1e6:.2f} µF')
for i, q in enumerate(Q_indiv):
    print(f'Carga Q{i+1}:              {q*1e3:.2f} mC')
print(f'Carga Total Qt:        {Qt_paralelo*1e3:.2f} mC')
```

```python
# 3. Resolução de Sistema de Malhas de Kirchhoff (Aula 04)
# Malha 1: 12 = 300*I1 + 100*I2
# Malha 2:  5 = 100*I1 + 200*I2
R_mat = np.array([
    [300.0, 100.0],
    [100.0, 200.0]
])
V_vet = np.array([12.0, 5.0])
I_correntes = np.linalg.solve(R_mat, V_vet)

print('Resolução das Leis de Kirchhoff (Exemplo Aula 04):')
print(f'Corrente de Malha I1: {I_correntes[0]*1e3:.2f} mA')
print(f'Corrente de Malha I2: {I_correntes[1]*1e3:.2f} mA')
print(f'Corrente no Ramo Central I3 (I1+I2): {(I_correntes[0]+I_correntes[1])*1e3:.2f} mA')
```
